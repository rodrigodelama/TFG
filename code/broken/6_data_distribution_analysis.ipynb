```python
# Create comprehensive comparison plots

def create_metric_comparison_plots(data):
    """Create comprehensive plots comparing all models and configurations"""
    
    # Set up the figure with subplots
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    fig.suptitle('Model Performance Comparison Across All Configurations', fontsize=16, fontweight='bold')
    
    metrics = ['mae', 'mse', 'r2']
    metric_names = ['Mean Absolute Error (MAE)', 'Mean Squared Error (MSE)', 'R² Score']
    
    # Plot 1: MAE comparison by percentile
    ax1 = axes[0, 0]
    for model_config in data['model_config'].unique():
        subset = data[data['model_config'] == model_config].sort_values('percentile')
        ax1.plot(subset['percentile'], subset['mae'], marker='o', label=model_config, linewidth=2)
    
    ax1.set_title('MAE by Percentile', fontweight='bold')
    ax1.set_xlabel('Percentile')
    ax1.set_ylabel('Mean Absolute Error')
    ax1.grid(True, alpha=0.3)
    ax1.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    
    # Plot 2: MSE comparison by percentile
    ax2 = axes[0, 1]
    for model_config in data['model_config'].unique():
        subset = data[data['model_config'] == model_config].sort_values('percentile')
        ax2.plot(subset['percentile'], subset['mse'], marker='o', label=model_config, linewidth=2)
    
    ax2.set_title('MSE by Percentile', fontweight='bold')
    ax2.set_xlabel('Percentile')
    ax2.set_ylabel('Mean Squared Error')
    ax2.grid(True, alpha=0.3)
    ax2.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    
    # Plot 3: R² comparison by percentile
    ax3 = axes[1, 0]
    for model_config in data['model_config'].unique():
        subset = data[data['model_config'] == model_config].sort_values('percentile')
        ax3.plot(subset['percentile'], subset['r2'], marker='o', label=model_config, linewidth=2)
    
    ax3.set_title('R² Score by Percentile', fontweight='bold')
    ax3.set_xlabel('Percentile')
    ax3.set_ylabel('R² Score')
    ax3.grid(True, alpha=0.3)
    ax3.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    
    # Plot 4: Box plot comparison of R² scores
    ax4 = axes[1, 1]
    data.boxplot(column='r2', by='model_config', ax=ax4, rot=45)
    ax4.set_title('R² Score Distribution by Model Configuration', fontweight='bold')
    ax4.set_xlabel('Model Configuration')
    ax4.set_ylabel('R² Score')
    
    plt.tight_layout()
    plt.show()

# Call the function 
# create_metric_comparison_plots(all_data)

# call the function for 99th percentile and lower
create_metric_comparison_plots(all_data[all_data['percentile'] <= 99])
```

```python
# Create improvement analysis

def analyze_improvements(data):
    """Analyze how feature engineering affects each model"""
    
    improvement_data = []
    
    for model in data['model'].unique():
        model_data = data[data['model'] == model]
        baseline = model_data[model_data['configuration'] == 'Baseline']
        fe = model_data[model_data['configuration'] == 'Feature Engineering']
        
        for percentile in baseline['percentile'].unique():
            if percentile in fe['percentile'].values:
                baseline_row = baseline[baseline['percentile'] == percentile].iloc[0]
                fe_row = fe[fe['percentile'] == percentile].iloc[0]
                
                # Calculate percentage improvements (negative means worse performance)
                mae_improvement = ((baseline_row['mae'] - fe_row['mae']) / baseline_row['mae']) * 100
                mse_improvement = ((baseline_row['mse'] - fe_row['mse']) / baseline_row['mse']) * 100
                r2_improvement = ((fe_row['r2'] - baseline_row['r2']) / abs(baseline_row['r2'])) * 100
                worst_avg_error_improvement = ((baseline_row['worst_avg_error'] - fe_row['worst_avg_error']) / baseline_row['worst_avg_error']) * 100
                
                # # Calculate percentage improvements on AVERAGE performance
                # mae_improvement = ((baseline['mae'].mean() - fe['mae'].mean()) / baseline['mae'].mean()) * 100
                # mse_improvement = ((baseline['mse'].mean() - fe['mse'].mean()) / baseline['mse'].mean()) * 100
                # r2_improvement = ((fe['r2'].mean() - baseline['r2'].mean()) / abs(baseline['r2'].mean())) * 100
                # worst_avg_error_improvement = ((baseline['worst_avg_error'].mean() - fe['worst_avg_error'].mean()) / baseline['worst_avg_error'].mean()) * 100
                improvement_data.append({
                    'model': model,
                    'percentile': percentile,
                    'mae_improvement': mae_improvement,
                    'mse_improvement': mse_improvement,
                    'r2_improvement': r2_improvement,
                    'worst_avg_error_improvement': worst_avg_error_improvement
                })
    
    improvement_df = pd.DataFrame(improvement_data)
    
    # Filter for 99th percentile and lower
    improvement_df = improvement_df[improvement_df['percentile'] <= 99]
    
    # Plot improvements in 2x2 layout
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    fig.suptitle('Feature Engineering Impact (≤99th Percentile): Percentage Improvement over Baseline', fontsize=16, fontweight='bold')
    
    metrics = ['mae_improvement', 'mse_improvement', 'r2_improvement', 'worst_avg_error_improvement']
    titles = ['MAE Improvement (%)', 'MSE Improvement (%)', 'R² Improvement (%)', 'ES-like AVG of 5% Worst Errors (%)']
    
    for i, (metric, title) in enumerate(zip(metrics, titles)):
        ax = axes[i//2, i%2]
        for model in improvement_df['model'].unique():
            model_data = improvement_df[improvement_df['model'] == model].sort_values('percentile')
            ax.plot(model_data['percentile'], model_data[metric], marker='o', label=model, linewidth=2)
        
        ax.axhline(y=0, color='red', linestyle='--', alpha=0.7, label='No improvement')
        ax.set_title(title, fontweight='bold')
        ax.set_xlabel('Percentile')
        ax.set_ylabel('Improvement (%)')
        ax.set_xticks(np.arange(75, 100, 5))
        ax.grid(True, alpha=0.3)
        ax.legend()
    
    plt.tight_layout()
    plt.show()
    
    return improvement_df

improvement_analysis = analyze_improvements(all_data)
```

```python
# Create separate distribution plots for each percentile

def create_percentile_distribution_plots(data):
    """Create separate distribution plots for each percentile"""
    
    percentiles = sorted(data['percentile'].unique())
    n_percentiles = len(percentiles)
    
    # Calculate grid dimensions
    n_cols = 3  # 3 columns for the 3 metrics
    n_rows = n_percentiles
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 4*n_rows))
    fig.suptitle('Metric Distribution by Percentile (≤99th Percentile)', fontsize=16, fontweight='bold')
    
    metrics = ['mae', 'mse', 'r2']
    metric_names = ['Mean Absolute Error', 'Mean Squared Error', 'R² Score']
    
    for row, percentile in enumerate(percentiles):
        percentile_data = data[data['percentile'] == percentile]
        
        for col, (metric, name) in enumerate(zip(metrics, metric_names)):
            ax = axes[row, col] if n_rows > 1 else axes[col]
            
            # Create box plot for this percentile and metric
            model_configs = percentile_data['model_config'].unique()
            values = [percentile_data[percentile_data['model_config'] == config][metric].values 
                     for config in model_configs]
            
            bp = ax.boxplot(values, labels=model_configs, patch_artist=True)
            
            # Color the boxes
            colors = plt.cm.Set3(np.linspace(0, 1, len(model_configs)))
            for patch, color in zip(bp['boxes'], colors):
                patch.set_facecolor(color)
                patch.set_alpha(0.7)
            
            ax.set_title(f'{name} - {percentile}th Percentile', fontweight='bold')
            ax.set_ylabel(name)
            ax.tick_params(axis='x', rotation=45)
            ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

create_percentile_distribution_plots(all_data_filtered)
```

```python
# Create violin plots for better distribution visualization

def create_percentile_violin_plots(data):
    """Create violin plots showing distribution shape for each percentile"""
    
    metrics = ['mae', 'mse', 'r2']
    metric_names = ['Mean Absolute Error', 'Mean Squared Error', 'R² Score']
    
    fig, axes = plt.subplots(1, 3, figsize=(20, 8))
    fig.suptitle('Value Distribution by Percentile - Violin Plots (≤99th Percentile)', fontsize=16, fontweight='bold')
    
    for i, (metric, name) in enumerate(zip(metrics, metric_names)):
        ax = axes[i]
        
        # Prepare data for violin plot
        percentiles = sorted(data['percentile'].unique())
        positions = range(len(percentiles))
        
        violin_data = []
        for percentile in percentiles:
            percentile_values = data[data['percentile'] == percentile][metric].values
            violin_data.append(percentile_values)
        
        # Create violin plot
        parts = ax.violinplot(violin_data, positions=positions, showmeans=True, showmedians=True)
        
        # Customize violin plot
        for pc in parts['bodies']:
            pc.set_facecolor('lightblue')
            pc.set_alpha(0.7)
        
        ax.set_title(f'{name} Distribution by Percentile', fontweight='bold')
        ax.set_xlabel('Percentile')
        ax.set_ylabel(name)
        ax.set_xticks(positions)
        ax.set_xticklabels([f'{p}th' for p in percentiles])
        ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

create_percentile_violin_plots(all_data_filtered)
```

```python
# Create heatmaps showing metric values across percentiles and configurations

def create_percentile_heatmaps(data):
    """Create heatmaps showing metric values across percentiles and configurations"""
    
    metrics = ['mae', 'mse', 'r2']
    metric_names = ['Mean Absolute Error', 'Mean Squared Error', 'R² Score']
    
    fig, axes = plt.subplots(1, 3, figsize=(20, 6))
    fig.suptitle('Metric Values Heatmap: Percentiles vs Model Configurations (≤99th Percentile)', fontsize=16, fontweight='bold')
    
    for i, (metric, name) in enumerate(zip(metrics, metric_names)):
        ax = axes[i]
        
        # Create pivot table
        pivot_data = data.pivot_table(values=metric, index='percentile', columns='model_config', aggfunc='mean')
        
        # Create heatmap
        im = ax.imshow(pivot_data.values, cmap='viridis', aspect='auto')
        
        # Set ticks and labels
        ax.set_xticks(range(len(pivot_data.columns)))
        ax.set_xticklabels(pivot_data.columns, rotation=45, ha='right')
        ax.set_yticks(range(len(pivot_data.index)))
        ax.set_yticklabels([f'{int(p)}th' for p in pivot_data.index])
        
        ax.set_title(f'{name}', fontweight='bold')
        ax.set_xlabel('Model Configuration')
        ax.set_ylabel('Percentile')
        
        # Add colorbar
        cbar = plt.colorbar(im, ax=ax)
        cbar.set_label(name)
        
        # Add value annotations
        for y in range(len(pivot_data.index)):
            for x in range(len(pivot_data.columns)):
                value = pivot_data.iloc[y, x]
                if not np.isnan(value):
                    ax.text(x, y, f'{value:.3f}', ha='center', va='center', 
                           color='white' if value < pivot_data.values.mean() else 'black')
    
    plt.tight_layout()
    plt.show()

create_percentile_heatmaps(all_data_filtered)
```

```python
# Create detailed statistics table for each percentile

def create_percentile_statistics_table(data):
    """Create detailed statistics for each percentile"""
    
    print("=" * 120)
    print("DETAILED STATISTICS BY PERCENTILE (≤99th PERCENTILE)")
    print("=" * 120)
    
    percentiles = sorted(data['percentile'].unique())
    metrics = ['mae', 'mse', 'r2']
    
    for percentile in percentiles:
        print(f"\n🎯 {percentile}th PERCENTILE:")
        print("-" * 80)
        
        percentile_data = data[data['percentile'] == percentile]
        
        for metric in metrics:
            print(f"\n   {metric.upper()}:")
            
            # Overall statistics
            overall_stats = percentile_data[metric].describe()
            print(f"      Overall: Mean={overall_stats['mean']:.4f}, Std={overall_stats['std']:.4f}, "
                  f"Min={overall_stats['min']:.4f}, Max={overall_stats['max']:.4f}")
            
            # By configuration
            for config in ['Baseline', 'Feature Engineering']:
                config_data = percentile_data[percentile_data['configuration'] == config]
                if not config_data.empty:
                    config_stats = config_data[metric].describe()
                    print(f"      {config}: Mean={config_stats['mean']:.4f}, Std={config_stats['std']:.4f}")
            
            # Best and worst performing models
            best_idx = percentile_data[metric].idxmax() if metric == 'r2' else percentile_data[metric].idxmin()
            worst_idx = percentile_data[metric].idxmin() if metric == 'r2' else percentile_data[metric].idxmax()
            
            best_model = percentile_data.loc[best_idx, 'model_config']
            best_value = percentile_data.loc[best_idx, metric]
            worst_model = percentile_data.loc[worst_idx, 'model_config']
            worst_value = percentile_data.loc[worst_idx, metric]
            
            print(f"      Best: {best_model} ({best_value:.4f})")
            print(f"      Worst: {worst_model} ({worst_value:.4f})")

create_percentile_statistics_table(all_data_filtered)
```

```python
# Analyze parameter impact on metrics

def analyze_parameter_impact(data):
    """Analyze how different parameters affect the error metrics"""
    
    print("Available columns in the dataset:")
    print(data.columns.tolist())
    print("\nData types:")
    print(data.dtypes)
    print("\nFirst few rows:")
    print(data.head())
    
    # Identify potential parameter columns (excluding metrics and identifiers)
    metric_cols = ['mae', 'mse', 'r2']
    identifier_cols = ['model', 'configuration', 'model_config', 'percentile']
    
    parameter_cols = [col for col in data.columns if col not in metric_cols + identifier_cols]
    
    print(f"\nPotential parameter columns: {parameter_cols}")
    
    return parameter_cols

parameter_columns = analyze_parameter_impact(all_data_filtered)
```

```python
# Create correlation analysis between parameters and metrics

def create_parameter_correlation_analysis(data, parameter_cols):
    """Create correlation analysis between parameters and metrics"""
    
    if not parameter_cols:
        print("No parameter columns found. Let's analyze the categorical variables we have:")
        
        # Analyze categorical variables
        categorical_analysis = {}
        
        # Model type impact
        model_impact = data.groupby('model')[['mae', 'mse', 'r2']].agg(['mean', 'std'])
        print("\n📊 IMPACT BY MODEL TYPE:")
        print(model_impact)
        
        # Configuration impact
        config_impact = data.groupby('configuration')[['mae', 'mse', 'r2']].agg(['mean', 'std'])
        print("\n🔧 IMPACT BY CONFIGURATION:")
        print(config_impact)
        
        # Percentile impact
        percentile_impact = data.groupby('percentile')[['mae', 'mse', 'r2']].agg(['mean', 'std'])
        print("\n📈 IMPACT BY PERCENTILE:")
        print(percentile_impact)
        
        return
    
    # If we have numerical parameters, create correlation matrix
    metrics = ['mae', 'mse', 'r2']
    analysis_cols = parameter_cols + metrics
    
    # Calculate correlation matrix
    correlation_matrix = data[analysis_cols].corr()
    
    # Create correlation heatmap
    plt.figure(figsize=(12, 8))
    sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
                square=True, fmt='.3f')
    plt.title('Parameter-Metric Correlation Matrix', fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()
    
    # Print strongest correlations with metrics
    print("\n🔍 STRONGEST CORRELATIONS WITH METRICS:")
    for metric in metrics:
        print(f"\n{metric.upper()} correlations:")
        metric_corr = correlation_matrix[metric].drop(metric).abs().sort_values(ascending=False)
        for param, corr in metric_corr.head(5).items():
            direction = "📈" if correlation_matrix[metric][param] > 0 else "📉"
            print(f"   {direction} {param}: {correlation_matrix[metric][param]:.3f}")

create_parameter_correlation_analysis(all_data_filtered, parameter_columns)
```

```python
# Create interaction effects analysis

def analyze_interaction_effects(data):
    """Analyze interaction effects between model type and configuration"""
    
    print("=" * 80)
    print("INTERACTION EFFECTS ANALYSIS")
    print("=" * 80)
    
    metrics = ['mae', 'mse', 'r2']
    
    for metric in metrics:
        print(f"\n📊 {metric.upper()} ANALYSIS:")
        print("-" * 40)
        
        # Create interaction analysis
        interaction_data = data.pivot_table(values=metric, 
                                          index='model', 
                                          columns='configuration', 
                                          aggfunc=['mean', 'std'])
        
        print("Mean values:")
        print(interaction_data['mean'])
        print("\nStandard deviation:")
        print(interaction_data['std'])
        
        # Calculate improvement for each model
        print(f"\nFeature Engineering Impact on {metric.upper()}:")
        for model in data['model'].unique():
            model_data = data[data['model'] == model]
            baseline_mean = model_data[model_data['configuration'] == 'Baseline'][metric].mean()
            fe_mean = model_data[model_data['configuration'] == 'Feature Engineering'][metric].mean()
            
            if metric == 'r2':
                improvement = ((fe_mean - baseline_mean) / abs(baseline_mean)) * 100
                direction = "↗️" if improvement > 0 else "↘️"
            else:
                improvement = ((baseline_mean - fe_mean) / baseline_mean) * 100
                direction = "↗️" if improvement > 0 else "↘️"
            
            print(f"   {model}: {direction} {improvement:+.2f}%")

analyze_interaction_effects(all_data_filtered)
```

```python
# Statistical significance testing

def perform_statistical_tests(data):
    """Perform statistical tests to determine significant parameter effects"""
    
    from scipy import stats
    
    print("=" * 80)
    print("STATISTICAL SIGNIFICANCE TESTS")
    print("=" * 80)
    
    metrics = ['mae', 'mse', 'r2']
    
    for metric in metrics:
        print(f"\n📊 {metric.upper()} TESTS:")
        print("-" * 40)
        
        # Test 1: Model type effect (ANOVA)
        model_groups = [data[data['model'] == model][metric].values for model in data['model'].unique()]
        f_stat, p_value = stats.f_oneway(*model_groups)
        print(f"Model type effect (ANOVA): F={f_stat:.4f}, p={p_value:.4f}")
        if p_value < 0.05:
            print("   ✅ Significant difference between models")
        else:
            print("   ❌ No significant difference between models")
        
        # Test 2: Configuration effect (T-test)
        baseline_values = data[data['configuration'] == 'Baseline'][metric].values
        fe_values = data[data['configuration'] == 'Feature Engineering'][metric].values
        t_stat, p_value = stats.ttest_ind(baseline_values, fe_values)
        print(f"Configuration effect (T-test): t={t_stat:.4f}, p={p_value:.4f}")
        if p_value < 0.05:
            print("   ✅ Significant difference between configurations")
        else:
            print("   ❌ No significant difference between configurations")
        
        # Test 3: Percentile correlation
        correlation, p_value = stats.pearsonr(data['percentile'], data[metric])
        print(f"Percentile correlation: r={correlation:.4f}, p={p_value:.4f}")
        if p_value < 0.05:
            print("   ✅ Significant correlation with percentile")
        else:
            print("   ❌ No significant correlation with percentile")

perform_statistical_tests(all_data_filtered)
```
